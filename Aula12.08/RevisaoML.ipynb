# Exemplo:
Criar um modelo preditivo para previsão da potência ativa (Global_active_power) usando como dados de entrada um dataframe com os atributos: 'Voltage', 'Global_intensity', 'Sub_metering_1'

```python
# Dados de entrada ===> Features:
# X ===> dataframe
X = dados[['Voltage', 'Global_intensity', 'Sub_metering_1']]
X.head()
```

```python
# Dados de saída ===> Target:
# y ===> Array
y = dados['Global_active_power']
y.head()
```

```python
# Separar os dados de X e y em treinamento e teste
# Treino ===> 70%
# Teste  ===> 30%
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.3)
```
